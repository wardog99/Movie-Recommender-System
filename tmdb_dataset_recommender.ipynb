```python
import numpy as np
import pandas as pd

#getting the path for the dataset
import os
for dirname, _, filenames in os.walk('/Users/partha/AI_ML_Projects/Recommender_System/datasets'):
    for f in filenames:
        print(os.path.join(dirname, f))
```

```python
#loading the dataset
movies = pd.read_csv('/Users/partha/AI_ML_Projects/Recommender_System/datasets/archive/tmdb_5000_movies.csv', sep=',')
credits = pd.read_csv('/Users/partha/AI_ML_Projects/Recommender_System/datasets/archive/tmdb_5000_credits.csv', sep=',')
```

```python
#lets merge the two datasets on the title column to get a single dataset with all the information
movies = movies.merge(credits,on='title')
```

```python
#taking important columns from the dataset to work with
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.dropna(inplace=True)
```

```python
import ast

def convert(text):
    L=[]
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
    
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
movies['cast']=movies['cast'].apply(convert)
```

```python
movies.head()
```

```python
movies['cast']=movies['cast'].apply(lambda x:x[0:3])
movies['cast']
```

```python
def fetch_director(text):
    L=[]
    for i in ast.literal_eval(text):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L

movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
def collapse(L):
    L1=[]
    for i in L:
        L1.append(i.replace(" ",""))
    return L1

movies['cast']=movies['cast'].apply(collapse)
movies['crew']=movies['crew'].apply(collapse)
movies['genres']=movies['genres'].apply(collapse)
movies['keywords']=movies['keywords'].apply(collapse)
```

```python
movies.head()
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies['tags']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
new = movies.drop(columns=['overview','genres','keywords','cast','crew'],axis=1)
new.head()
```

```python
new['tags']=new['tags'].apply(lambda x: " ".join(x))
new.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features =5000, stop_words = 'english')

vector = cv.fit_transform(new['tags']).toarray()

vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
new[new['title']=='The Lego Movie'].index[0]
```

```python
def recommend(movie):
    index = new[new['title']==movie].index[0]
    distances = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda x:x[1])
    for i in distances[1:20]:
        print(new.iloc[i[0]].title+ " --------- Cosine Similarity: "+str(i[1]))

recommend('The Dark Knight Rises')
```
